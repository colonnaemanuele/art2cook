# Deepchecks Dataset Evaluation

Given that our `task` involves `Generative AI`, we decided to adopt a different organization for the images folder, treating our image dataset as if it were part of a `classification task`.

First, we need to load the dataset.

```python
import os
from deepchecks.vision.checks import ImageDatasetDrift
from deepchecks.vision.suites import train_test_validation
from deepchecks.vision import classification_dataset_from_directory
```

```python
current_dir = os.getcwd()
data_dir = os.path.join(current_dir, "data")
imgs_dir = os.path.join(data_dir, "deepchecks_images")

print(f'Current directory: {current_dir}\n'
      f'Data directory: {data_dir}\n'
      f'Images directory: {imgs_dir}')
```

```python
train_ds, test_ds = classification_dataset_from_directory(
    root=imgs_dir, object_type='VisionData', image_extension='jpg')
```

With this setup in place, we tested our dataset using [**ImageDatasetDrift**](https://docs.deepchecks.com/0.13/checks_gallery/vision/train_test_validation/plot_image_dataset_drift.html) to evaluate whether the images in the `train set` are similar to those in the `test set` based on specific properties such as aspect ratio, area, brightness, and more.

```python
check = ImageDatasetDrift()
result = check.run(train_dataset=train_ds, test_dataset=test_ds)
result
```

Lastly, we run the [**train_test_validation suite**](https://docs.deepchecks.com/stable/tabular/auto_tutorials/quickstarts/plot_quick_train_test_validation.html).

```python
suite = train_test_validation()
result = suite.run(train_ds, test_ds)
result.show()
```
